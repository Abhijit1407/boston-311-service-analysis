```python
# Load the data
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

```python
file_path = "/Workspace/Users/more.ab@northeastern.edu/ALY6110_Mod 4 Final Project_Basic Analysis/311 Service Requests_2025.csv"
```

```python
# Read the CSV
df = spark.read.csv(file_path, header=True, inferSchema=True)
```

```python
# Display basic info
print(f"Total Records: {df.count()}")
print(f"Total Columns: {len(df.columns)}")
```

```python
display(df.limit(10))
```

```python
# Check the schema
df.printSchema()
```

```python
# Check for null values in key columns
from pyspark.sql.functions import col, count, when, isnan

null_counts = df.select([
    count(when(col(c).isNull(), c)).alias(c) 
    for c in df.columns
])

print("Null value counts:")
display(null_counts)
```

```python
# Basic statistics on key columns
print("\nCase Status Distribution:")
display(df.groupBy("case_status").count().orderBy(col("count").desc()))

print("\nTop 10 Request Types:")
display(df.groupBy("type").count().orderBy(col("count").desc()).limit(10))

print("\nTop 10 Departments:")
display(df.groupBy("department").count().orderBy(col("count").desc()).limit(10))
```

```python
# Geographic Clustering: Which neighborhoods generate what types of requests?

print("=== INSIGHT 1: Geographic Clustering of Service Request Types ===\n")

# Requests by neighborhood
neighborhood_analysis = df.groupBy("neighborhood", "type") \
    .count() \
    .orderBy(col("count").desc())

print("Top Request Types by Neighborhood:")
display(neighborhood_analysis.limit(20))

# Top neighborhoods by total requests
top_neighborhoods = df.groupBy("neighborhood") \
    .count() \
    .orderBy(col("count").desc()) \
    .limit(10)

print("\nTop 10 Neighborhoods by Total Requests:")
display(top_neighborhoods)

# Request type distribution by neighborhood (pivot for clustering view)
neighborhood_type_pivot = df.groupBy("neighborhood") \
    .pivot("type") \
    .count() \
    .fillna(0)

print("\nNeighborhood-Type Matrix (first 10 neighborhoods):")
display(neighborhood_type_pivot.limit(10))
```

```python
# Calculate resolution time and analyze performance

print("=== INSIGHT 2: Response Time Performance Correlation ===\n")

from pyspark.sql.functions import datediff, avg, percentile_approx

# Calculate resolution time in days
df_with_resolution = df.filter(col("closed_dt").isNotNull()) \
    .withColumn("resolution_days", 
                datediff(col("closed_dt"), col("open_dt")))

# Average resolution time by department
dept_performance = df_with_resolution.groupBy("department") \
    .agg(
        count("*").alias("total_requests"),
        avg("resolution_days").alias("avg_resolution_days"),
        percentile_approx("resolution_days", 0.5).alias("median_resolution_days")
    ) \
    .orderBy(col("avg_resolution_days").desc())

print("Department Performance - Average Resolution Time:")
display(dept_performance)

# On-time performance by department
ontime_by_dept = df.groupBy("department", "on_time") \
    .count() \
    .orderBy("department", col("count").desc())

print("\nOn-Time Performance by Department:")
display(ontime_by_dept)

# Resolution time by request type
type_performance = df_with_resolution.groupBy("type") \
    .agg(
        count("*").alias("total_requests"),
        avg("resolution_days").alias("avg_resolution_days")
    ) \
    .orderBy(col("avg_resolution_days").desc()) \
    .limit(15)

print("\nAverage Resolution Time by Request Type:")
display(type_performance)
```

```python
# Temporal patterns analysis

print("=== INSIGHT 3: Temporal and Seasonal Patterns ===\n")

from pyspark.sql.functions import month, dayofweek, hour, date_format

# Add temporal columns
df_temporal = df.withColumn("month", month(col("open_dt"))) \
    .withColumn("day_of_week", dayofweek(col("open_dt"))) \
    .withColumn("month_name", date_format(col("open_dt"), "MMMM"))

# Monthly trends
monthly_requests = df_temporal.groupBy("month", "month_name") \
    .count() \
    .orderBy("month")

print("Monthly Request Volume (2025):")
display(monthly_requests)

# Day of week patterns
dow_requests = df_temporal.groupBy("day_of_week") \
    .count() \
    .orderBy("day_of_week")

print("\nRequests by Day of Week (1=Sunday, 7=Saturday):")
display(dow_requests)

# Source channel analysis
source_analysis = df.groupBy("source") \
    .count() \
    .orderBy(col("count").desc())

print("\nRequest Submission Source:")
display(source_analysis)

# Seasonal patterns by request type
seasonal_types = df_temporal.groupBy("month", "month_name", "type") \
    .count() \
    .orderBy("month", col("count").desc())

print("\nSeasonal Request Type Patterns:")
display(seasonal_types.limit(30))
```

```python
from pyspark.sql.functions import datediff, month, dayofweek, date_format, year, col

# Add calculated columns
df_export = df.withColumn("resolution_days", 
    datediff(col("closed_dt"), col("open_dt"))) \
    .withColumn("month", month(col("open_dt"))) \
    .withColumn("month_name", date_format(col("open_dt"), "MMMM")) \
    .withColumn("day_of_week", dayofweek(col("open_dt"))) \
    .withColumn("year", year(col("open_dt")))

# Display first to verify
display(df_export.limit(10))

# Then create a temp view and download
df_export.createOrReplaceTempView("main_export")
```
